```python
# find the second largest no. in the list

l1 = [3,11,51,12,45]

max =float("-inf")
sec_max = float("-inf")+

for i in l1:
  if i>max:
    sec_max =max
    max = i
  elif i>sec_max and max>i:
    sec_max =i

print(sec_max)
```

```python
# sort a dict by key

students = [
    {"name": "Amit", "marks": 75},
    {"name": "Neha", "marks": 90},
    {"name": "Raj", "marks": 60}
]

result = sorted(students,key=lambda x:x["marks"])

print(result)
```

```python
#map , filter, reduce

from functools import reduce

l = [1,2,3,4,5]
l1 =map(lambda x:x*2, l)
l2 =filter(lambda x:x%2==0,l)
l3 =reduce(lambda a,b:a+b,l)
print(list(l1))
print(list(l2))
print(l3)
```

```python
# group items into even and odd
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

result = {"even": [], "odd": []}

for i in numbers:
  if i%2==0:
    result["even"].append(i)
  else:
    result["odd"].append(i)

print(result)
```

```python
# class __init__ method and inhertence

class Car:
  def __init__(self,brand,year):
    self.brand=brand
    self.year=year

  def display(self):
    print(f"car brand is {self.brand}, year of realse is {self.year}")

class Child(Car):
  pass

c1 = Child("bmw",2026)
c1.display()
```
